Category Analysis

```python
# CATEGORY SALES ANALYSIS


category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

print("Category-wise Sales")
print(category_sales)

print("\n")



# REGION SALES ANALYSIS


region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)

print("Region-wise Sales")
print(region_sales)

print("\n")



# CATEGORY PROFIT ANALYSIS


category_profit = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)

print("Category-wise Profit")
print(category_profit)

print("\n")



# MONTHLY SALES TREND


monthly_sales = df.groupby('Order Month')['Sales'].sum()

print("Monthly Sales Trend")
print(monthly_sales)

print("\n")



# AVERAGE DISCOUNT


average_discount = df['Discount'].mean()

print("Average Discount Given:", round(average_discount, 2))
```

```python
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
category_profit = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)

print("Category Sales:\n", category_sales)
print("\nCategory Profit:\n", category_profit)
```

Category Performance Insight

- Technology generates the highest sales and profit contribution.
- Furniture shows high sales but extremely low profit.
- Office Supplies remains stable in both sales and profit.

**Conclusion:** Furniture category has margin inefficiency despite strong revenue contribution. ##

```python
subcat_profit = df.groupby('Sub-Category')[['Sales','Profit']].sum().sort_values('Profit')

print(subcat_profit)
```

Sub-Category Profit Insight

- Copiers, Phones, and Accessories are top profit contributors.
- Tables and Bookcases are loss-making sub-categories.
- Some products generate sales but reduce overall profitability.

**Conclusion:** Certain sub-categories are major profit leak points and need pricing/discount correction.

```python
discount_profit = df.groupby('Discount')[['Sales','Profit']].mean()
corr = df[['Discount','Profit','Sales']].corr()

print("Discount vs Profit:\n", discount_profit)
print("\nCorrelation:\n", corr)
```

Discount Impact Insight

- Higher discount levels (30%+) lead to negative profit.
- Low discount orders generate healthy margins.
- Strong negative correlation between discount and profit.

**Conclusion:** Discount strategy is reducing profitability and needs stricter control.

```python
top_customers = df.groupby('Customer ID')[['Sales','Profit']].sum().sort_values('Sales', ascending=False).head(10)

customer_share = df.groupby('Customer ID')['Sales'].sum()
top_share = customer_share.sort_values(ascending=False).head(10) / customer_share.sum()

print("Top Customers:\n", top_customers)
print("\nRevenue Share:\n", top_share)
```

Customer Insight

- A small group of customers contributes significant revenue.
- Top customers show high dependency in overall revenue structure.
- Revenue distribution is uneven across customer base.

**Conclusion:** Business is dependent on a small set of high-value customers (risk concentration).

```python
region_sales = df.groupby('Region')[['Sales','Profit']].sum().sort_values('Profit')

state_loss = df.groupby('State')[['Sales','Profit']].sum().sort_values('Profit').head(10)

print("Region Sales & Profit:\n", region_sales)
print("\nTop Loss States:\n", state_loss)
```

Region Performance Insight

- West and East regions are highly profitable.
- Central region underperforms in profitability.
- States like Texas and Ohio show consistent losses.

**Conclusion:** Geographic imbalance exists between revenue and profitability.

```python
monthly_sales = df.groupby('Order Month')['Sales'].sum()
monthly_profit = df.groupby('Order Month')['Profit'].sum()
yearly = df.groupby('Order Year')[['Sales','Profit']].sum()

print("Monthly Sales:\n", monthly_sales)
print("\nMonthly Profit:\n", monthly_profit)
print("\nYearly Trend:\n", yearly)
```

Time Trend Insight

- Sales and profit peak during November and December.
- Early months show weaker performance.
- Overall business shows consistent year-on-year growth.

**Conclusion:** Strong seasonal pattern exists with peak dependency on year-end demand.
